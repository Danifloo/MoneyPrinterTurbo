# GoNews — Video de Artemis II (Google Colab)

Un video vertical **1080×1920, 30 fps** para TikTok, Instagram Reels y YouTube Shorts, con **narración en español latino, fotos reales de NASA, música instrumental y sin subtítulos**.

**Pulsa ▶ en la única celda de código.** El programa descargará material oficial de NASA, generará la voz y montará el MP4. No requiere claves API de MoneyPrinterTurbo, Pexels ni ngrok. El video se guarda en `/content/Artemis_II_GoNews_9x16.mp4` y al terminar aparece un enlace para descargarlo.

**Fuentes informativas:** [NASA — Artemis II](https://www.nasa.gov/mission/artemis-ii/), [NASA — Amerizaje 10 de abril](https://www.nasa.gov/blogs/missions/2026/04/10/artemis-ii-flight-day-10-re-entry-live-updates/). Consulta los [términos de uso de imágenes NASA](https://www.nasa.gov/nasa-brand-center/images-and-media/).

**Nota:** Google Colab necesita acceso a internet y puede suspender las sesiones gratuitas. Las voces Edge TTS pueden fallar; se usa una alternativa si eso sucede. Si no se encuentran fotografías verdaderas o si falla el audio, se detiene sin fabricar imágenes. Guarda el MP4 antes de cerrar Colab.


```python
# CREA EL MP4 EN GOOGLE COLAB. No necesita claves API.
import subprocess, sys
from pathlib import Path
from urllib.request import urlretrieve
from IPython.display import display, Video, FileLink

print("Preparando herramientas de video y voz...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q",
                "edge-tts", "gTTS", "pillow", "numpy", "requests"], check=True)

script_path = Path("/content/crear_video_artemis2.py")
urlretrieve(
    "https://raw.githubusercontent.com/Danifloo/MoneyPrinterTurbo/main/docs/crear_video_artemis2.py",
    script_path
)
print("Creando video con fotografías reales de NASA, voz y música...")
subprocess.run([sys.executable, str(script_path)], check=True)

mp4 = Path("/content/Artemis_II_GoNews_9x16.mp4")
assert mp4.exists() and mp4.stat().st_size>100000, "No se creó el archivo de video."
print("✅ Listo. Reproduce o descarga tu video:")
display(Video(str(mp4), embed=False, width=360))
display(FileLink(str(mp4)))
print("Puedes guardarlo en tu Samsung con el enlace de arriba.")
```
